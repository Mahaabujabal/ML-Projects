```python
df= pd.read_csv('tmdb_movies_data.csv')
df.head()
```

```python
#practicing on the question ( whats is the number of realesed movies annualy? )
#1 showing a specific column line-chart 
movies_Count = pd.DataFrame(df.release_year.value_counts(ascending=True))
movies_Count
```

```python
#2 adding an index to the previus output
movies_Count= movies_Count.reset_index()
movies_Count
```

```python
#3 Changing name of columns & sorting years from smaller to larger:
movies_Count.columns=['year','count']
movies_Count= movies_Count.sort_values(by=['year'])
movies_Count
```

```python
#4 Finally showing the plot for movie counting
plt.plot(movies_Count['year'],movies_Count['count']);
```

```python
plt.plot(movies_Count['year'],movies_Count['count'], c='0.2'); #changing line color
plt.xlabel('Years')
plt.ylabel('Movies Count')
plt.title('Number of Realesed Movies Annually')
plt.xlim(1960,2015); #controlling where data starts and ends
plt.xticks(range(960,2016, 5));#controlling where data starts and ends & the jump between each number
plt.axis('off');
```

```python
plt.plot(movies_Count["year"], movies_Count["count"], c="red")
plt.plot(
    range(1960, 2015, 10),
    [500, 400, 300, 200, 100, 600],
    c="green",
    linestyle=":",
    linewidth=4,
)
plt.xlabel("Years")
plt.ylabel("Movies Count")
plt.title("Number of Released Movies Annually")
plt.legend (['Movies','Series'])
```

```python
df= pd.read_csv('tmdb_movies_data.csv')
df.head()
#Question ماهي العلاقة بين معدل التصويت والارباح؟
```

```python
plt.scatter(df.vote_average, df.revenue)
plt.xlabel('Vote_Average')
plt.ylabel('Revenue')
plt.title('Relation between Vote Average and Revenue')
```

```python
#Modifying on scatter_chart
plt.scatter(df.vote_average, df.revenue, alpha = 0.7, marker= '*', s= 150, label = 'Vote Average' )
plt.scatter(df.vote_count, df.revenue, alpha = 0.8, marker= '+', s= 150, label = 'Vote Count')
plt.legend(loc= 'upper right')
plt.xlabel('Vote_Average & Count')
plt.ylabel('Revenue')
plt.title('Relation between Vote Average & Count vs.Revenue')
```
