```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tips= sns.load_dataset("tips")
tips.head ()
```

```python
diamonds = sns.load_dataset('diamonds')
diamonds.head()
```

```python
diamonds.cut.value_counts(normalize=True)
```

```python
sns.histplot(data= diamonds, x='price', bins=10)
```

```python
sns.histplot(data= tips, x='total_bill', bins=10)
```

```python
sns.kdeplot(data= tips, x='total_bill')
```

```python
sns.displot(data= tips, x='total_bill', kind='kde')
```

```python
sns.countplot(data=diamonds, x="cut", hue="cut", palette="bright", legend=False)
```

```python
plt.pie(tips.smoker.value_counts(), labels=['Non_Smoker','Smoker'], autopct='%1.2f%%')
```

```python
tips.smoker.value_counts
```

```python
plt.pie(tips.smoker.value_counts(), labels= tips.smoker.value_counts().index, autopct='%1.2f%%')
```

```python
#Binariative data Numerical x Numerical (line,scatter) charts
sns.lineplot(data=tips, x='size',y='tip')
```

```python
sns.scatterplot(data=tips, x='size',y='tip')
```

```python
#Categorical x Categorical values:
sns.barplot(data=tips, x='day',y='total_bill', hue='sex')
```

```python
sns.histplot(diamonds.sort_values('clarity').reset_index(), x='clarity', hue='cut',alpha=0.3)
```

```python
sns.histplot(diamonds.sort_values('clarity').reset_index(), x='clarity', hue='cut',alpha=0.7,multiple='dodge')
```

```python
#Numirical x Categorical
sns.barplot(data=diamonds, x='color', y='price', hue='color' )
```

```python
sns.violinplot(data=diamonds, x='color', y='price', hue='color' )
```
